```python
import plotly
import plotly.graph_objs as go
plotly.offline.init_notebook_mode()
import numpy as np
import matplotlib.pyplot as plt
import os
import json
```

```python
# finds all the json paths
# assumes all files are stored in this folder
basePath = "../db_files/db_files"

jsonPaths = np.array([])
first50JsonPaths = np.array([])

for number_folder in os.listdir(basePath):
    if int(number_folder) > 50: continue
    for jsonList in os.listdir(f"{basePath}/{number_folder}"):
        if jsonList.endswith(".json"):
            first50JsonPaths = np.append(first50JsonPaths, f"{basePath}/{number_folder}/{jsonList}")
            
jsonPaths = np.sort(jsonPaths)
```

```python
# from json paths finds all numbers/POSCAR paths with a HSE

# appends all valid HSE to these lists
numbers = []
validPaths = []
validJsonPaths = []

for file in jsonPaths:
    with open(file, "r") as f:
        data = json.load(f)
        for number in ["-2", "-1", "0", "1", "2"]:
            if (data.get(number) is not None and data[number].get("hse-gs") is not None):
                rawNum = file.split("/")[-1].split("_")[0]
                validJsonPaths.append(file)
                pathHSE = f'{basePath}/{rawNum}/{number}/{rawNum}_hse-gs_POSCAR'
                validPaths.append(pathHSE)
                numbers.append(number)
```
